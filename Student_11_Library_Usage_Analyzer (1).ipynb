## 4. NumPy practice

```python

# TODO: Select numeric columns relevant to your project
numeric_cols = df[['Books_Issued', 'Books_Returned', 'Late_Returns']].to_numpy
# TODO: Convert them to a NumPy array using .to_numpy()
num_array = numeric_cols.to_numpy()
# TODO: Calculate average, maximum and minimum
average_vals = np.mean(num_array, axis=0)
max_vals = np.max(num_array, axis=0)
min_vals = np.min(num_array, axis=0)
print("Average (Issued, Returned, Late):", average_vals.tolist())
print("Maximum (Issued, Returned, Late):", max_vals.tolist())
print("Minimum (Issued, Returned, Late):", min_vals.tolist())


```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
from IPython.display import display
# TODO: Create calculated column(s)
df['Return_Rate'] = (df['Books_Returned'] / df['Books_Issued']) * 100
df['Unreturned_Books'] = df['Books_Issued'] - df['Books_Returned']

# TODO: Filter records using a condition
high_late_returns = df[df['Late_Returns'] > 2]
print("Students with > 2 late returns:")
display(high_late_returns)
# TODO: Sort records and display the top/bottom results
sorted_by_late = df.sort_values(by='Late_Returns', ascending=False)
print("Top 5 students with most late returns:")
display(sorted_by_late.head(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate return rate; identify students with most late returns; summarise usage.

```python
# TODO: Solve your assigned challenge
# 1. Calculate Return Rate (%) and Unreturned Books
df['Return_Rate'] = (df['Books_Returned'] / df['Books_Issued']) * 100
df['Unreturned_Books'] = df['Books_Issued'] - df['Books_Returned']

# 2. Identify top 5 students with most late returns
top_late_returns = df.sort_values(by='Late_Returns', ascending=False).head(5)
print("Top 5 students with most late returns:")
display(top_late_returns)

# 3. Summarize overall usage
total_issued = df['Books_Issued'].sum()
total_returned = df['Books_Returned'].sum()
avg_return_rate = df['Return_Rate'].mean()
total_late = df['Late_Returns'].sum()

print("\n--- Library Usage Summary ---")
print(f"Total Books Issued: {total_issued}")
print(f"Total Books Returned: {total_returned}")
print(f"Average Return Rate: {avg_return_rate:.2f}%")
print(f"Total Late Returns: {total_late}")
```
